```python
# Step 1: Imports
import numpy as np
import soundfile as sf
from scipy.io import loadmat
from scipy.signal import fftconvolve
import os
```

```python
# Step 2: Load WAV Files
channels = {
    "FrontLeft": "surround_sound/FrontLeft.wav",
    "FrontRight": "surround_sound/FrontRight.wav",
    "Center": "surround_sound/Center.wav",
    "RearLeft": "surround_sound/RearLeft.wav",
    "RearRight": "surround_sound/RearRight.wav"
}
```

```python
audio_data = {}
for name, path in channels.items():
    data, sr = sf.read(path)
    audio_data[name] = data
```

```python
hrtf = loadmat("surround_sound/hrir_final.mat")
```

```python
# CIPIC indexing: (azimuth, elevation)
# Angles mapped from your layout:
angles = {
    "FrontLeft": (-30, 0),
    "FrontRight": (30, 0),
    "Center": (0, 0),
    "RearLeft": (-110, 0),
    "RearRight": (110, 0)
}
```

```python
from scipy.io import loadmat

hrtf = loadmat('surround_sound/hrir_final.mat')  # or whichever subject you have
print(hrtf.keys())
```

```python
# CIPIC standard azimuth and elevation grids
azimuths = np.array([
    -80, -65, -55, -45, -40, -35, -30, -25, -20, -15,
    -10, -5, 0, 5, 10, 15, 20, 25, 30, 35,
    40, 45, 55, 65, 80
])

elevations = np.linspace(-45, 230.625, 50)

# Match the closest indices for a desired (azimuth, elevation) pair
def get_hrtf_indices(az_deg, el_deg):
    az_idx = (np.abs(azimuths - az_deg)).argmin()
    el_idx = (np.abs(elevations - el_deg)).argmin()
    return el_idx, az_idx
```

```python
# Step 4: Convolve Each Channel with Corresponding HRTFs
def apply_hrtf(signal, hrtf_l, hrtf_r):
    left = fftconvolve(signal, hrtf_l, mode='full')[:len(signal)]
    right = fftconvolve(signal, hrtf_r, mode='full')[:len(signal)]
    return np.vstack((left, right))
```

```python
vss_mix = np.zeros((2, len(audio_data["FrontLeft"])))
```

```python
for name, signal in audio_data.items():
    az, el = angles[name]
    el_idx, az_idx = get_hrtf_indices(az, el)

    h_l = hrtf['hrir_l'][el_idx, az_idx, :]
    h_r = hrtf['hrir_r'][el_idx, az_idx, :]

    stereo = apply_hrtf(signal, h_l, h_r)
    vss_mix[:, :len(stereo[0])] += stereo
```

```python
# Normalize VSS output
vss_mix /= np.max(np.abs(vss_mix))
sf.write("output_vss.wav", vss_mix.T, sr)
```

```python
# Step 5: Simple Stereo Downmix (No HRTF)
L = audio_data["FrontLeft"]
R = audio_data["FrontRight"]
C = audio_data["Center"]
RL = audio_data["RearLeft"]
RR = audio_data["RearRight"]
```

```python
# Step: Ensure all signals are same length (trim to shortest)
min_len = min([len(sig) for sig in audio_data.values()])

for key in audio_data:
    audio_data[key] = audio_data[key][:min_len]

# Now safely unpack trimmed signals
L = audio_data["FrontLeft"]
R = audio_data["FrontRight"]
C = audio_data["Center"]
RL = audio_data["RearLeft"]
RR = audio_data["RearRight"]
```

```python
# Common downmix coefficients
downmix_L = 0.3254 * L + 0.2301 * C + 0.2301 * RL
downmix_R = 0.3254 * R + 0.2301 * C + 0.2301 * RR
```

```python
stereo_downmix = np.vstack((downmix_L, downmix_R))
stereo_downmix /= np.max(np.abs(stereo_downmix))
sf.write("output_downmix.wav", stereo_downmix.T, sr)

print("✅ Done! Files saved: output_vss.wav and output_downmix.wav")
```
